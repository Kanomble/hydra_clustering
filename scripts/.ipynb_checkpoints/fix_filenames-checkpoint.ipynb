# This script was utilized to rename RSEM result files to sample descriptions

```python
import os
```

```python
def change_filenames(directory:str):
    files = os.listdir(directory)
    for file in files:
        splitted_file = file.split("_S")
        new_file_header = splitted_file[0]
        if "genes" in splitted_file[1]:
            new_file_ending = ".genes.results"
        elif "isoforms" in splitted_file[1]:
            new_file_ending = ".isoforms.results"
        else:
            print("[*] ERROR")

        new_file = new_file_header + new_file_ending
        os.rename(directory+file,directory+new_file)
    print("[*] DONE")
```

```python
eco_dir = "../data/rsem_counts_final/EcoKD1/"
ahl_dir = "../data/rsem_counts_final/HydraAHL/"
rec_dir = "../data/rsem_counts_final/HydraRecolonization/"
temp_dir = "../data/rsem_counts_final/HydraTemperature/"

change_filenames(ahl_dir)
change_filenames(rec_dir)
change_filenames(temp_dir)
```

```python
for file in os.listdir(eco_dir):
    splitted_file = file.split(".")[0].split("_S")[0] + ".genes.results"
    old_file = eco_dir + file
    new_file = eco_dir + splitted_file
    os.rename(old_file,new_file)
```

```python
for file in os.listdir(ahl_dir):
    new_file = file.replace("-L","-S")
    new_file = ahl_dir + new_file
    old_file = ahl_dir + file
    os.rename(old_file,new_file)
```

```python
for file in os.listdir(rec_dir):
    new_file = file.replace("-L","-S")
    new_file = rec_dir + new_file
    old_file = rec_dir + file
    os.rename(old_file,new_file)
```

```python
for file in os.listdir(temp_dir):
    if "L1" in file:
        new_file = file.replace("-L1","")
    elif "L2" in file:
        new_file = file.replace("-L2","")
    else:
        new_file = file
    new_file = temp_dir + new_file
    old_file = temp_dir + file
    os.rename(old_file,new_file)
```
