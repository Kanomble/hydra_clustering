```python
import pandas as pd
```

```python
popfile = "../data/goatools_data/population.txt"
population_genes = []
with open(popfile,"r") as pfile:
    for line in pfile.readlines():
        population_genes.append(line.rstrip())
```

```python
df = pd.read_csv("../results/deseq2_rsem_sortmerna/tables/EcoKD1_Eco1KD_B8_vs_control_B8_results.csv")
df["gene"] = df["Unnamed: 0"].apply(lambda x: x.split(".")[1])
df.head()
```

```python
sample = "../results/network_analysis/genes_module_green.txt"
samplefile = "../results/network_analysis/goatools_ecokd1_sample_genes.txt"
sample_genes = []
with open(sample,"r") as sfile:
    with open(samplefile,"w") as sgenes:
        for line in sfile.readlines():
            gene = line.split(".")[1]
            gene = gene.rstrip()
            if gene in population_genes:
                if gene in df["gene"].to_list():
                    # print(gene)
                    sample_genes.append(gene)
                    sgenes.write(gene+"\n")
```

```python
df[df["gene"].isin(sample_genes)].sort_values(by="log2FoldChange")
```
